```python
import seaborn as sns
import matplotlib.pyplot as plt

data = df.select_dtypes(include='number')

correlation = data.corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```
