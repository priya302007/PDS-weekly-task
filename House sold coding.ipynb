```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")

df["date"].head()
```

```python
df["Sale_Year"] = df["date"].dt.year

df[["date", "Sale_Year"]].head()
```

```python
for col in ["bedrooms", "bathrooms"]:
    df[col] = df[col].replace(0, np.nan)
    df[col] = df[col].fillna(df[col].median())

print(df[["bedrooms", "bathrooms"]].head(10))
```

```python
# Select numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

# Select categorical columns
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Display output
print(df.isnull().sum())
```

```python
df["Property_Age"] = df["Sale_Year"] - df["yr_built"]

df[["yr_built", "Sale_Year", "Property_Age"]].head()
```

```python
df["Is_Renovated"] = np.where(
    df["yr_renovated"] > 0, "Yes", "No"
)

df[["yr_renovated", "Is_Renovated"]].head()
```

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["price"] < lower_bound) |
    (df["price"] > upper_bound)
]

print("Number of Outliers:", len(outliers))
```
